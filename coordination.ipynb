Вывод:
Анализ 5 worst-case кадров показал, что модель Depth Anything v2 корректно восстанавливает общий глобальный тренд глубины сцены, однако существенно сглаживает локальную геометрию и теряет детали на границах объектов. Наибольшие ошибки наблюдаются в дальних и структурно сложных областях — деревьях, столбах, границах автомобилей и линии горизонта. Карты ошибок показывают, что depth prediction плохо справляется с резкими depth discontinuities и тонкими структурами.

Все worst-case примеры относятся к последовательному фрагменту Scene02 (frames 00108–00113), что указывает на систематическую, а не случайную ошибку модели. Вероятной причиной является domain gap между реальными изображениями, на которых обучалась monocular depth model, и синтетическими сценами VKITTI2. Несмотря на корректное восстановление общей структуры сцены, модель демонстрирует нестабильность на синтетических текстурах на дальних объектах и высокочастотных деталях.

**[стартовое]** Используем pre-trained Depth Anything v2 Small (без дообучения), YOLOv8n (pretrained на COCO), Lift-to-3D через калибровочную матрицу из KITTI. _Обоснование:_ ограничение по времени (2 недели) и ресурсам (free Colab).

**[стартовое]** Класс для фокуса — Car (больше всего примеров в KITTI). Pedestrian/Cyclist — опционально, если успеем.
