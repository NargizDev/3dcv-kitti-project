```python
from google.colab import drive
drive.mount('/content/drive')

import json
import random
from pathlib import Path
```

```python
BASE = Path('/content/drive/MyDrive/3dcv-project')

CONFIG_DIR = BASE / 'results' / 'track_b' / 'configs'

config_paths = {
    'real_only': CONFIG_DIR / 'real_only.json',
    'synth_only': CONFIG_DIR / 'synth_only.json',
    'mixed_50_50': CONFIG_DIR / 'mixed_50_50.json'
}

for name, path in config_paths.items():
    print(name, '->', path.exists(), path)
```

```python
validation_results = {}

for name, path in config_paths.items():

    with open(path, 'r') as f:
        data = json.load(f)

    files = data['files']

    missing = [
        file_path
        for file_path in files
        if not Path(file_path).exists()
    ]

    validation_results[name] = {
        'declared_count': data['count'],
        'actual_count': len(files),
        'missing_count': len(missing),
        'missing_examples': missing[:10]
    }

    print(f'\n{name}')
    print('Declared count:', data['count'])
    print('Actual count  :', len(files))
    print('Missing files :', len(missing))

    if len(missing) > 0:
        print('Examples:')
        for m in missing[:10]:
            print(m)
```

```python
random.seed(42)

for name, path in config_paths.items():

    with open(path, 'r') as f:
        data = json.load(f)

    files = data['files']

    sample_size = min(10, len(files))
    sampled_files = random.sample(files, sample_size)

    print(f'\n{name} — random sample ({sample_size} files)')

    for file_path in sampled_files:
        exists = Path(file_path).exists()
        print(f'{exists} -> {file_path}')
```

```python
print('\n====================================================')
print('VALIDATION SUMMARY')
print('====================================================')

all_ok = True

for name, result in validation_results.items():

    print(f'\n{name}')
    print(f'Declared count : {result["declared_count"]}')
    print(f'Actual count   : {result["actual_count"]}')
    print(f'Missing files  : {result["missing_count"]}')

    if result['declared_count'] != result['actual_count']:
        all_ok = False
        print('Count mismatch')

    if result['missing_count'] > 0:
        all_ok = False
        print('Some files are missing')

if all_ok:
    print('\n✅ All configs are valid')
else:
    print('\n⚠️ Some problems were found')
```
